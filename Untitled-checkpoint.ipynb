```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import pickle
import glob

# --- ADIM 1: KLASÖRDEKİ CSV DOSYALARINI OTOMATİK BUL VE BİRLEŞTİR ---
print("Klasördeki CSV dosyaları aranıyor...")
csv_dosyalari = glob.glob("*.csv")

if len(csv_dosyalari) < 3:
    print(f"HATA: Klasörde en az 3 CSV dosyası olmalı. Bulunan dosya sayısı: {len(csv_dosyalari)}")
    print(f"Klasör içeriği: {csv_dosyalari}")
else:
    print(f"Bulunan ve birleştirilecek dosyalar: {csv_dosyalari[:3]}")
    
    # İlk 3 CSV dosyasını otomatik oku ve birleştir
    df_listesi = [pd.read_csv(dosya) for dosya in csv_dosyalari[:3]]
    df = pd.concat(df_listesi, ignore_index=True)

    print("\nVeriler başarıyla birleştirildi!")
    print("Sütunların orijinal isimleri:", df.columns.tolist())

    # --- ADIM 2: SÜTUN İSİMLERİNİ OTOMATİK EŞLEŞTİR ---
    # Kodun hata vermemesi için sütun isimlerini temizliyoruz
    df.columns = df.columns.str.strip().str.lower()
    
    # Yaygın kullanılan tarih ve yoğunluk sütun adlarını kontrol et
    tarih_adaylari = ['tarih', 'date', 'date_time', 'datetime', 'zaman']
    yogunluk_adaylari = ['yogunluk', 'traffic_index', 'density', 'number_of_vehicles', 'index']
    
    tarih_sutunu = next((c for c in df.columns if c in tarih_adaylari), df.columns[0])
    yogunluk_sutunu = next((c for c in df.columns if c in yogunluk_adaylari), df.columns[1])
    
    print(f"Kullanılacak Tarih Sütunu: '{tarih_sutunu}'")
    print(f"Kullanılacak Yoğunluk Sütunu: '{yogunluk_sutunu}'")

    # Temizlik işlemleri
    df = df.dropna(subset=[yogunluk_sutunu])
    df[tarih_sutunu] = pd.to_datetime(df[tarih_sutunu])
    df = df.sort_values(by=tarih_sutunu).reset_index(drop=True)

    # --- ADIM 3: ÖZNİTELİK MÜHENDİSLİĞİ ---
    print("\nModele zaman özellikleri ekleniyor...")
    df['saat'] = df[tarih_sutunu].dt.hour
    df['haftanin_gunu'] = df[tarih_sutunu].dt.dayofweek
    df['hafta_sonu'] = df['haftanin_gunu'].apply(lambda x: 1 if x >= 5 else 0)
    df['bir_saat_once'] = df[yogunluk_sutunu].shift(1)
    df = df.dropna().reset_index(drop=True)

    # --- ADIM 4: VERİ BÖLME ---
    X = df[['saat', 'haftanin_gunu', 'hafta_sonu', 'bir_saat_once']]
    y = df[yogunluk_sutunu]
    bolme_noktasi = int(len(df) * 0.8)
    X_train, X_test = X.iloc[:bolme_noktasi], X.iloc[bolme_noktasi:]
    y_train, y_test = y.iloc[:bolme_noktasi], y.iloc[bolme_noktasi:]

    # --- ADIM 5: MODEL EĞİTİMİ ---
    print("Model eğitiliyor...")
    model = xgb.XGBRegressor(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=5,
        tree_method='hist',
        random_state=42
    )
    model.fit(X_train, y_train)

    # --- ADIM 6: PERFORMANS TESTİ ---
    print("\nEğitim bitti! Model test ediliyor...")
    tahminler = model.predict(X_test)
    mae = mean_absolute_error(y_test, tahminler)
    r2 = r2_score(y_test, tahminler)

    print(f"\n--- SINAV SONUÇLARI ---")
    print(f"Ortalama Hata Payı (MAE): {mae:.2f}")
    print(f"Başarı Oranı (R2 Score): % {r2*100:.2f}")

    with open('trafik_modeli.pkl', 'wb') as dosya:
        pickle.dump(model, dosya)
    print("\nModel başarıyla 'trafik_modeli.pkl' adıyla kaydedildi!")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import pickle
import glob

print("Klasördeki CSV dosyaları aranıyor...")
csv_dosyalari = glob.glob("*.csv")

print(f"Bulunan dosyalar birleştiriliyor: {csv_dosyalari[:3]}")
df_listesi = [pd.read_csv(dosya) for dosya in csv_dosyalari[:3]]
df = pd.concat(df_listesi, ignore_index=True)

# Sütun isimlerini standart hale getir
df.columns = df.columns.str.strip().str.upper()

# Eksik verileri temizle ve tarihi formatla
df = df.dropna(subset=['NUMBER_OF_VEHICLES'])
df['DATE_TIME'] = pd.to_datetime(df['DATE_TIME'])

print("İstanbul geneli için saatlik bazda veri gruplanıyor (Hızlandırma ve Doğruluk Hamlesi)...")
# Aynı saatteki tüm koordinatların araç sayılarını toplayarak İstanbul geneli tek bir zaman serisi elde ediyoruz
df_genel = df.groupby('DATE_TIME')['NUMBER_OF_VEHICLES'].sum().reset_index()
df_genel = df_genel.sort_values(by='DATE_TIME').reset_index(drop=True)

print("Modele zaman özellikleri ekleniyor...")
df_genel['saat'] = df_genel['DATE_TIME'].dt.hour
df_genel['haftanin_gunu'] = df_genel['DATE_TIME'].dt.dayofweek
df_genel['hafta_sonu'] = df_genel['haftanin_gunu'].apply(lambda x: 1 if x >= 5 else 0)
df_genel['bir_saat_once'] = df_genel['NUMBER_OF_VEHICLES'].shift(1)

df_genel = df_genel.dropna().reset_index(drop=True)

print("Veri eğitim ve test olarak bölünüyor...")
X = df_genel[['saat', 'haftanin_gunu', 'hafta_sonu', 'bir_saat_once']]
y = df_genel['NUMBER_OF_VEHICLES']

bolme_noktasi = int(len(df_genel) * 0.8)
X_train, X_test = X.iloc[:bolme_noktasi], X.iloc[bolme_noktasi:]
y_train, y_test = y.iloc[:bolme_noktasi], y.iloc[bolme_noktasi:]

print("Model yeniden eğitiliyor...")
model = xgb.XGBRegressor(
    n_estimators=150,
    learning_rate=0.08,
    max_depth=6,
    tree_method='hist',
    random_state=42
)
model.fit(X_train, y_train)

print("\nEğitim bitti! Yeni model test ediliyor...")
tahminler = model.predict(X_test)
mae = mean_absolute_error(y_test, tahminler)
r2 = r2_score(y_test, tahminler)

print(f"\n--- YENİ SINAV SONUÇLARI ---")
print(f"Ortalama Hata Payı (MAE): {mae:.2f} araç")
print(f"Başarı Oranı (R2 Score): % {r2*100:.2f}")

with open('trafik_modeli.pkl', 'wb') as dosya:
    pickle.dump(model, dosya)
print("\nGeliştirilmiş model 'trafik_modeli.pkl' adiyla kaydedildi!")
```
